# Logo

```python
import numpy as np
import svgwrite
from svg2png import svg2png
```

```python
def logo(dwg, center=(0,0), size=10, linewidth=1, stroke="black", fill="none"):
    """Logo du sinus cardinal dans un cercle."""
    
    # Sinus cardinal
    N = 70 # Nombre de points
    x = center[0] + np.linspace(-size/2, size/2, N)
    n = np.linspace(-5, 5, N)
    y = center[1] + size/10 - np.sinc(n) * size/3

    # Liste des coordonnées du sinus cardinal
    points = ""
    for k in range(N):
        p = "M" if k==0 else "L"
        points += f"{p} {x[k]} {y[k]}"

    # Masque
    mask = dwg.mask(id="themask")
    mask.add( dwg.circle(center=center, r=size/2) )
    # dwg.defs.add(mask)

    # Cercle
    dwg.add( dwg.circle(center=center, r=size/2, stroke=stroke, fill=fill, stroke_width=linewidth) )

    # Dessin du sinus cardinal
    dwg.add( dwg.path(d=points, stroke=stroke, fill="none", stroke_width=linewidth) )

    return dwg
```

```python
# Logo site
W, H = 50, 50
dwg = svgwrite.Drawing('logo-signal2-site.svg', size=(f'{W}px', f'{H}px'))
dwg = logo(dwg, center=(W/2, H/2), size=.96*H, linewidth=1.7, stroke="black", fill="white")
dwg.save()
```

```python
# Logo Moodle
filename = 'logo-signal2-moodle'
W, H = 540, 190
dwg = svgwrite.Drawing(f'{filename}.svg', size=(f'{W}px', f'{H}px'))
gradient = dwg.linearGradient((0, 0), (1, 0), id="gradient")
gradient.add_stop_color(offset='0%',   color='rgb(0, 176, 50)', opacity=None)
gradient.add_stop_color(offset='100%', color='rgb(0, 219, 51)', opacity=None)
dwg.defs.add(gradient)
dwg.add(
    dwg.rect(insert=(0, 0), size=(f'{W}', f'{H}'), fill="url(#gradient)")
)
dwg = logo(dwg, center=(W/2, H/2), size=.8*H, linewidth=6, stroke="white", fill="none")
dwg.save()
svg2png(f'{filename}.svg')
```

```python
# Taille de l'image
W, H = 50, 50

# Nombre de points du sinus cardinal
N = 70

# Sinus cardinal
n = np.linspace(-1, 1, N)
x = np.linspace(0, W, N)
y = np.sinc(5*n)
y = 30 - y*15

# Liste des coordonnées du sinus cardinal
points = ""
for k in range(N):
    if k==0:
        points += "M "
    else:
        points += "L "
    points += f"{x[k]} {y[k]}"

# Initialisation du SVG
dwg = svgwrite.Drawing('logo-signal2-site.svg', size=(f'{W}px', f'{H}px'))

# # Rectangle blanc
# dwg.add( dwg.rect(insert=(0, 0), size=(f'{W}', f'{H}'), fill="red") )

# Masque
mask = dwg.mask(id="themask")
mask.add(
    dwg.circle(center=(W/2, H/2), r=.48*W, fill="white")
)
dwg.defs.add(mask)

# Cercle
dwg.add(
    dwg.circle(center=(W/2, H/2), r=.48*W, stroke="black", fill="white", stroke_width=1.5)
)

# Dessin du sinus cardinal
dwg.add(
    dwg.path(d=points, stroke="black", fill="none", stroke_width=2, mask="url(#themask)")
)

# Enregistrement
dwg.save()
```
