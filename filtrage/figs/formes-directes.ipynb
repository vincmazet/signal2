```python
fig, ax = blk.drawing(15, 5.8, debug=False)

# RIF ================================================================================

ox, oy = 1, 2.5

# Retard
blk.block(ax, 3-ox, 6-oy, 1, 1, "$z^{-1}$")

# Coefficients
blk.block(ax, 5-ox, 5-oy, 1, 1, "$1$")

# Sommateurs
blk.add(ax, 7.5-ox, 7.5-oy)

# Flèches horizontales (gauche)
blk.arrow(ax, 2.0-ox, 7.5-oy, 5, 0)
blk.arrow(ax, 3.5-ox, 5.5-oy, 1.5, 0)

# Flèches verticales (gauche)
blk.arrow(ax, 3.5-ox, 7.5-oy, 0, -.5)
blk.line(ax, 3.5-ox, 6.0-oy, 0, -.5)

# Flèches horizontales (droite)
blk.line(ax, 6-ox, 5.5-oy, 1.5, 0)

# Flèches verticales (droite)
blk.arrow(ax, 7.5-ox, 5.5-oy, 0, 1.5)

# Etiquettes
blk.label(ax,  1.5-ox, 7.5-oy, "$x[n]$")

# Milieu =============================================================================

blk.arrow(ax, 8-ox, 7.5-oy, 1, 0)

# RII ================================================================================

ox = -5

# Retard
blk.block(ax, 7-ox, 6-oy, 1, 1, "$z^{-1}$")
blk.block(ax, 7-ox, 4-oy, 1, 1, "$z^{-1}$")

# Coefficients
blk.block(ax, 5-ox, 5-oy, 1, 1, "$-2$")
blk.block(ax, 5-ox, 3-oy, 1, 1, "$-1$")

# Sommateurs
blk.add(ax, 3.5-ox, 7.5-oy)
blk.add(ax, 3.5-ox, 5.5-oy)

# Flèches horizontales (gauche)
blk.arrow(ax, 4-ox, 7.5-oy, 5, 0)
blk.arrow(ax, 5-ox, 5.5-oy, -1, 0)
blk.line(ax, 5-ox, 3.5-oy, -1.5, 0)

# Flèches verticales (gauche)
blk.arrow(ax, 3.5-ox, 6.0-oy, 0, 1)
blk.arrow(ax, 3.5-ox, 3.5-oy, 0, 1.5)

# Flèches horizontales (droite)
blk.arrow(ax, 7.5-ox, 5.5-oy, -1.5, 0)
blk.arrow(ax, 7.5-ox, 3.5-oy, -1.5, 0)

# # Flèches verticales (droite)
blk.arrow(ax, 7.5-ox, 7.5-oy, 0, -.5)
blk.arrow(ax, 7.5-ox, 6.0-oy, 0, -1)
blk.line( ax, 7.5-ox, 4.0-oy, 0, -0.5)

# Etiquettes
blk.label(ax,  9.5-ox, 7.5-oy, "$y[n]$")
blk.label(ax,  7.5, 0.5, "$H_1$")

# ====================================================================================

blk.saving(ax, "forme-directe-sys1.svg")
blk.saving(ax, "forme-directe-sys1.png")
```
