# Réponse impulsionnelle de $e^{-c t}$

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('../../src/signal.mplstyle')
```

```python
# Réponse impulsionnelle
def h(t, c):
    return np.where( t>=0, np.exp(-c*t), 0)

# Initialisation de l'affichage
colors = ['C0', 'C1', 'C2', 'C3']
fig = plt.figure(figsize=(6,4))

# Affichage
t = np.arange(-1, 6, .01)
cc = np.arange(.5, -.5, -.25)
for k in range(len(cc)):
    clr = colors[k]
    c = cc[k]
    if k==0:
        y = h(4,c) - .35
    elif k==3:
        y = h(4,c) - .15
    else:
        y = h(4,c) + .1
    plt.plot(t, h(t,c), color=clr)
    plt.text(4.02, y, f"$c={c}$", color=clr)

# Finalisation de l'affichage
plt.axis([-.5, 5, -.5, 3])
plt.xticks(range(5))
plt.yticks(range(3))

fig.savefig(f"ri-exp.svg", bbox_inches='tight')
```
