# Réponse impulsionnelle de $e^{-c t}$

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('../../src/signal.mplstyle')
```

```python
def h(t, c):
    return np.where( t>=0, np.exp(-c*t), 0)

t = np.arange(-1, 6, .01)
fig = plt.figure(figsize=(15,2))
i=0
for c in np.arange(.5, -.5, -.25):
    i+=1
    plt.subplot(1,4,i)
    plt.plot(t, h(t,c))
    plt.axis([-.5, 5, -.5, 3])
    plt.xticks(np.arange(0,5,1))
    plt.xlabel(f"$c={c}$", size=16)

fig.savefig(f"ri-exp.svg", bbox_inches='tight')
fig.savefig(f"ri-exp.png", bbox_inches='tight')
```
