# ROC de $a^n u[n]$

Transformée en Z :
$$
X[z] = \frac{1}{1-az^{-1}}
$$


Zéro en $z=0$, pôle en $z=a$

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
plt.style.use('../../src/signal.mplstyle')
```

```python
a = .7

fig, ax = plt.subplots(1, 1, figsize=(4,4))
ax.set_aspect('equal')
ax.set_axisbelow(True)

N = 3
ax.set_xticks(np.arange(-N, N+1, 1))
ax.set_xlim([-N, N])
ax.set_yticks(np.arange(-N, N+1, 1))
ax.set_ylim([-N, N])

# ROC
# https://stackoverflow.com/questions/22789356/plot-a-donut-with-fill-or-fill-between-use-pyplot-in-matplotlib
n = 50
lim = 2.2
radii = [a, lim*2]
theta = np.linspace(0, 2*np.pi, n, endpoint=True)
xs = np.outer(radii, np.cos(theta))
ys = np.outer(radii, np.sin(theta))
xs[1,:] = xs[1,::-1]
ys[1,:] = ys[1,::-1]
ax.fill(np.ravel(xs), np.ravel(ys), color='tab:blue', alpha=.2)
    
ax.plot(0, 0, marker='o', markersize=10, color='tab:blue', markeredgewidth=2, fillstyle='none')
ax.plot(a, 0, marker='x', markersize=10, color='tab:blue', markeredgewidth=2)

unit_circle = patches.Circle((0,0), radius=1, fill=False, color='tab:orange', lw=1.5)
ax.add_patch(unit_circle)
ax.annotate(text='', xytext=(1.1,-2), xy=(.5, -.8), arrowprops=dict(arrowstyle='->', color="tab:orange"))
ax.text(1.1, -2, "lieu de\nla TFTD", ha="left", va="center", color="tab:orange",
         bbox=dict(boxstyle="round", ec="tab:orange", fc=(1,1,1))
       )

ax.set_xlabel('$\\mathrm{Re}[z]$')
ax.set_ylabel('$\\mathrm{Im}[z]$')

plt.savefig("roc-z.svg")
plt.savefig("roc-z.png")
plt.show()
```
