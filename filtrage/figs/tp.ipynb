```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Figure
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 8)
ax.set_ylim(-2, 5)
ax.set_aspect("equal")
ax.axis("off")

# Initialisation des coordonnées
x = []
y = []

# Ligne de gauche
x += [0, 1]
y += [0, 0]

# P
n = np.linspace(1, 2, 100)
x += list(n)
y += list(-2 * (n-1.5) ** 2 + 0.5)

# Ligne du milieu
x += [2, 3, 3.5, 4, 4.5, 5, 6]
y += [0, 0, -0.5, 4, -1, 0, 0]

# S
n = np.linspace(6, 7, 100)
x += list(n)
y += list(-4 * (n-6.5) ** 2 + 1)

# Ligne de droite
x += [7, 8]
y += [0, 0]

# Affichage du tracé
ax.plot(x, y, color="black", linewidth=4)

# Étiquettes
labels = [
    [1.5, -0.5, "P"],
    [3.5, -1.5, "Q"],
    [4.0,  4.5, "R"],
    [4.5, -1.5, "S"],
    [6.5, -0.5, "T"],
]
for x, y, s in labels:
    ax.text(x, y, s, ha="center", va="center", size=30)

# Enregistrement
plt.savefig("ecg.svg", bbox_inches='tight')
# plt.show()
```
