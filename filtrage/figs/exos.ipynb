```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

plt.style.use('../../src/signal.mplstyle')
mpl.rcParams['lines.linewidth'] = 1.8
mpl.rcParams['axes.prop_cycle'] = mpl.cycler(color=['1f77b4'])
mpl.rcParams["axes.grid"] = False
```

```python
# Affichage
fig, ax = plt.subplots(1, 1, figsize=(7,4))
plt.plot([0, 100, 100], [-3, -3, -40])
plt.plot([0, 120, 120, 200], [0, 0, -30, -30])
plt.xlabel("$f$ (Hz)")
plt.ylabel("$20 \\log|H|$ (dB)")
plt.xlim([-10, 200])
plt.ylim([-40, 2])
plt.xticks([0, 100, 120])
plt.yticks([-30, -3, 0],[-30, -1, 0])
plt.savefig("figs-ex3.svg", bbox_inches="tight")
plt.savefig("figs-ex3.png", bbox_inches="tight")
plt.show()
```

```python
# Affichage
fig, ax = plt.subplots(1, 1, figsize=(7,4))

d = 4
plt.plot([0, 25-d], [-60, -60], '--', lw=1)
plt.plot([25-d, 25+d], [-60, -60])
plt.plot([25+d, 25+d, 200], [-60, 0, 0], '--', lw=1)
plt.plot([50, 50], [-90, -5], '--', lw=1)
plt.plot([50, 180], [-5, -5])
plt.plot([180, 200], [-5, -5], '--', lw=1)

plt.xlabel("$f$ (Hz)")
plt.ylabel("$20 \\log|H|$ (dB)")
plt.xlim([-1, 200])
plt.ylim([-90, 3])
plt.xticks([0, 25, 50, 175], [0, "2,5k", "50k", "2M"])
plt.yticks([-60, -5, 0],[-60, -3, 0])

plt.annotate(text='', xy=(25-d-2,-70), xytext=(25+d+2,-70), arrowprops=dict(arrowstyle='<->'))
plt.annotate("fréquence\nà éliminer", (25, -75), ha="center", va="top")
plt.annotate(text='', xy=(50,-70), xytext=(175,-70), arrowprops=dict(arrowstyle='<->'))
plt.annotate("bande utile", (112.5, -75), ha="center", va="top")

plt.savefig("figs-ex6.svg", bbox_inches="tight")
plt.savefig("figs-ex6.png", bbox_inches="tight")
plt.show()
```
