# ROC de $e^{-\delta t} u(t)$

Transformée de Laplace :
$$
X(s) = \frac{1}{s+\delta}
$$

Pas de zéro, pôle en $s = -\delta$

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
plt.style.use('../../src/signal.mplstyle')
```

```python
delta = 1.5

fig, ax = plt.subplots(1, 1, figsize=(4,4))
ax.set_aspect('equal')
ax.set_axisbelow(True)

N = 3
ax.set_xticks(np.arange(-N, N+1, 1))
ax.set_xlim([-N, N])
ax.set_yticks(np.arange(-N, N+1, 1))
ax.set_ylim([-N, N])

roc = patches.Rectangle((-delta,-N), width=2*N-delta, height=2*N, fill=True, color='tab:blue', alpha=0.2)
ax.add_patch(roc)
    
ax.plot(-delta, 0, marker='x', markersize=10, color='tab:blue', markeredgewidth=2)

ax.plot([0, 0],[-10, 10], color='tab:orange', linewidth=1.5)
ax.annotate(text='', xytext=(.8,-1.5), xy=(.1,-2), arrowprops=dict(arrowstyle='->', color="tab:orange"))
ax.text(.8, -1.5, "lieu de la\ntransformée\nde Fourier", ha="left", va="center", color="tab:orange",
         bbox=dict(boxstyle="round", ec="tab:orange", fc=(1,1,1))
       )

ax.set_xlabel('$\\sigma$')
ax.set_ylabel('$\\omega$')

plt.savefig("roc-laplace.svg")
plt.savefig("roc-laplace.png")
plt.show()
```
