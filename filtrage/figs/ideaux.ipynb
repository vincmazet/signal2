```python
import scipy.signal as sig

fn = 1
f = np.linspace(0,fn,200)

# Filtre idéal
fc = .4

# Gabarit
fcm, fcp = .33, .45
Am, Ap, att = .9, 1.1, .2
Amlog = 20*np.log10(Am)
Aplog = 20*np.log10(Ap)

# Réponse fréquentielle
E = np.exp(-(f-fc)*100)
G1 = (E + 0.1) / (1+E)
G2 = .1*np.sin(f*50)
G3 = np.exp(-(f-fc)**2/0.1)
G = G1 + np.multiply(G2,G3)
```

```python
# Sans échelle log
fig, ax = plt.subplots(1, 1, figsize=(8,6))
ax.axis('off')
plt.xlim([0, fn+.1])
plt.ylim([-.5, Ap+.1])

# Axes
plt.annotate(text='', xytext=(-.01, 0), xy=(fn+.1, 0), arrowprops=dict(arrowstyle='->', color="k"))
plt.annotate(text='', xytext=(0, -.01), xy=(0, Ap+.1), arrowprops=dict(arrowstyle='->', color="k"))

# Tracés
ax.fill_between([0, fcm, fcm, fcp, fcp, fn],
                [Am, Am, 0, 0, 0, 0],
                [Ap, Ap, Ap, Ap, att, att],
                color=(.17, .63, .17, .15))
plt.plot([0, fc, fc, fn], [1, 1, 0, 0], color="tab:blue")
plt.plot([0, fcp, fcp, fn], [Ap, Ap, att, att], linewidth=1.5, color="tab:green")
plt.plot([0, fcm, fcm, fn], [Am, Am, 0, 0], linewidth=1.5, color="tab:green")
plt.plot(f, G, color="tab:red")

# Textes
plt.text(-.02, Am, "$A^-$", ha="right", va="center")
plt.text(-.02, 1, "$K$", ha="right", va="center")
plt.text(-.02, Ap, "$A^+$", ha="right", va="center")
plt.text(fcm, -.04 , "$f_c-$", ha="right", va="top")
plt.text(fc , -.04 , "$f_c$", ha="center", va="top")
plt.text(fcp, -.04 , "$f_c+$", ha="left", va="top")
plt.text(-.02, -.04, "$0$", ha="right", va="top")
plt.text(fn+.1 , 0 , "$f$", ha="left", va="top")
plt.text(0 , Ap+.1, "$|H(f)|$", ha="right", va="bottom")
ax.annotate(text='', xytext=(fcm-.1,.1), xy=(fc,.1), arrowprops=dict(arrowstyle='->', color="tab:blue"))
ax.text(fcm-.1 , 0.1 , "filtre idéal", ha="right", va="center", fontsize=18, color="tab:blue")
ax.text(fn+.02 , 0.1 , "filtre réalisable", ha="left", va="center", fontsize=18, color="tab:red")

# Bandes
ax.annotate(text='', xytext=(0,-.2), xy=(fcm,-.2), arrowprops=dict(arrowstyle='<->'))
ax.text(fcm/2 , -.35 , "bande\npassante", ha="center", va="center", fontsize=18)
ax.annotate(text='', xytext=(fcm,-.2), xy=(fcp,-.2), arrowprops=dict(arrowstyle='<->'))
ax.text(fcm+(fcp-fcm)/2 , -.35 , "bande\nde transition", ha="center", va="center", fontsize=18)
ax.annotate(text='', xytext=(fcp,-.2), xy=(fn,-.2), arrowprops=dict(arrowstyle='<->'))
ax.text(fcp+(fn-fcp)/2 , -.35 , "bande\natténuée", ha="center", va="center", fontsize=18)

fig.savefig(f"filtre-reel.svg", bbox_inches='tight')
fig.savefig(f"filtre-reel.png", bbox_inches='tight')
plt.show()
```

```python
# Avec échelle log
fig, ax = plt.subplots(1, 1, figsize=(8,4))
ax.axis('off')
plt.xlim([0, fn+.1])
plt.ylim([-22, Aplog+2])

# Axes
plt.annotate(text='', xytext=(-.01, 0), xy=(fn+.1, 0), arrowprops=dict(arrowstyle='->', color="k"))
plt.annotate(text='', xytext=(0, -20), xy=(0, Aplog+2), arrowprops=dict(arrowstyle='->', color="k"))

# Tracés
ax.fill_between([0, fcm, fcm, fcp, fcp, fn],
                20*np.log10([Am, Am, .1, .1, .1, .1]),
                20*np.log10([Ap, Ap, Ap, Ap, att, att]),
                color=(.17, .63, .17, .15))
plt.plot([0, fc, fc], [0, 0, -20], color="tab:blue")
plt.plot([0, fcp, fcp, fn], 20*np.log10([Ap, Ap, att, att]), linewidth=1.5, color="tab:green")
plt.plot([0, fcm, fcm], [Amlog, Amlog, -20], linewidth=1.5, color="tab:green")
plt.plot(f, 20*np.log10(G), color="tab:red")
plt.plot([0, fn], [-22, -22], color="white", linewidth=35)

# Textes
# plt.text(-.02, Am, "$A^-$", ha="right", va="center")
# plt.text(-.02, 1, "$K$", ha="right", va="center")
# plt.text(-.02, Ap, "$A^+$", ha="right", va="center")
plt.text(fcm, 3, "$f_c-$", ha="right", va="top")
plt.text(fc , 3, "$f_c$", ha="center", va="top")
plt.text(fcp, 3, "$f_c+$", ha="left", va="top")
plt.text(-.02, 0, "$20\log_{10}K$", ha="right", va="center")
plt.text(fn+.1 , 0 , "$f$", ha="left", va="bottom")
plt.text(0 , Aplog+2, "$20\log_{10}|H(f)|$", ha="right", va="bottom")
ax.annotate(text='', xytext=(fcm-.1,-17), xy=(fc,-17), arrowprops=dict(arrowstyle='->', color="tab:blue"))
ax.text(fcm-.1 , -17 , "filtre idéal", ha="right", va="center", fontsize=18, color="tab:blue")
ax.text(fn-.15 , -17 , "filtre réalisable", ha="left", va="center", fontsize=18, color="tab:red")

# Bandes
ax.annotate(text='', xytext=(0,-21), xy=(fcm,-21), arrowprops=dict(arrowstyle='<->'))
ax.text(fcm/2 , -24 , "bande\npassante", ha="center", va="center", fontsize=18)
ax.annotate(text='', xytext=(fcm,-21), xy=(fcp,-21), arrowprops=dict(arrowstyle='<->'))
ax.text(fcm+(fcp-fcm)/2 , -24 , "bande\nde transition", ha="center", va="center", fontsize=18)
ax.annotate(text='', xytext=(fcp,-21), xy=(fn,-21), arrowprops=dict(arrowstyle='<->'))
ax.text(fcp+(fn-fcp)/2 , -24 , "bande\natténuée", ha="center", va="center", fontsize=18)

# Atténuations
ax.annotate(text='', xytext=(.53, Amlog-.7), xy=(.53,Aplog+.7), arrowprops=dict(arrowstyle='<->', color="tab:orange"))
ax.text(.55 , 1 , "ondulation en bande passante", ha="left", va="center", fontsize=18, color="tab:orange")
ax.annotate(text='', xytext=(.85, 0), xy=(.85,20*np.log10(att)), arrowprops=dict(arrowstyle='<->', color="tab:orange"))
ax.text(.84 , 20*np.log10(att)/2 , "atténuation", ha="right", va="center", fontsize=18, color="tab:orange")

fig.savefig(f"filtre-reel-log.svg", bbox_inches='tight')
fig.savefig(f"filtre-reel-log.png", bbox_inches='tight')
plt.show()
```
