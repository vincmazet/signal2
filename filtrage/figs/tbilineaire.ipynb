# Transformée bilinéaire

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
plt.style.use('../../src/signal.mplstyle')

import matplotlib as mpl
mpl.rcParams['font.size'] = 18
```

## Région de la transformée bilinéaire

```python
fig, axs = plt.subplots(1, 2, figsize=(9,4))

N = 3

# Plan en s
roc = patches.Rectangle((-N,-N), width=N, height=2*N, fill=True, color='tab:blue', alpha=0.2)
axs[0].add_patch(roc)
axs[0].set_xlim([-N, N])
axs[0].set_ylim([-N, N])  
axs[0].set_xlabel('$\\sigma = \\mathrm{Re}(s)$')
axs[0].set_ylabel('$\\omega = \\mathrm{Im}(s)$')
axs[0].set_aspect('equal')

# Plan en z
roc = patches.Circle((0,0), radius=1, fill=True, color='tab:blue', alpha=0.2)
axs[1].add_patch(roc)
axs[1].set_xlim([-N, N])
axs[1].set_ylim([-N, N])  
axs[1].set_xlabel('$\\mathrm{Re}(z)$')
axs[1].set_ylabel('$\\mathrm{Im}(z)$')
axs[1].set_aspect('equal')

plt.savefig("tbilineaire.svg")
plt.savefig("tbilineaire.png")
plt.show()
```

## Transformation

```python
def tb(s):
    k = 1
    if (1-s/k) != 0:
        z = (1+s/k)/(1-s/k)
    else:
        z = 0
    return z
```

```python
t = np.arange(0, 6.3, 0.05)
xx = np.cos(t)
yy = np.sin(t)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10,5))
axs = ax[0]
axz = ax[1]
axz.plot(xx,yy,"k")

A, B = 20, 20
d = 2

for a in np.arange(-A, d, d):
    for b in np.arange(-B, B+d, d):
        
        s = a + 1j*b
        z = tb(s)
        
        r = (a+A)/A
        g = (b+B)/(2*B)
        b = 1-g
        
        axs.plot(s.real, s.imag, marker=".", color=(r,g,b))
        axz.plot(z.real, z.imag, marker=".", color=(r,g,b))
        
axs.set_xlim([-A, A])
axs.set_ylim([-B, B])
axs.grid()
axz.set_xlim([-1.5, 1.5])
axz.set_ylim([-1.5, 1.5])
axz.grid()
```
