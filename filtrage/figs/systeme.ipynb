# Représentation d'un système en schéma-bloc

```python
import numpy as np
import blocks as blk
```

```python
fig, ax = blk.drawing(6, 1.2, debug=False)

y = .6

blk.label(ax, 0.5 , y    , "Entrée\n$x$")
blk.arrow(ax, 1.1 , y    , 0.9, 0)
blk.block(ax, 2.0 , y-0.5, 2, 1, "Système\n$\\mathcal{S}$")
blk.arrow(ax, 4.0 , y    , 0.9, 0)
blk.label(ax, 5.5 , y    , "Sortie\n$y$")

blk.saving(ax, "systeme1.svg")
blk.saving(ax, "systeme1.png")
```

```python
fig, ax = blk.drawing(6, 1.2, debug=False)

y = .6

blk.label(ax, 0.5 , y+.2 , "Entrée")
blk.label(ax, 0.5 , y-.2 , "$x=\\delta$")
blk.arrow(ax, 1.1 , y    , 0.9, 0)
blk.block(ax, 2.0 , y-0.5, 2, 1, "Système\n$\\mathcal{S}$")
blk.arrow(ax, 4.0 , y    , 0.9, 0)
blk.label(ax, 5.5 , y+.2 , "Sortie")
blk.label(ax, 5.5 , y-.2 , "$y=h$")

blk.saving(ax, "systeme2.svg")
blk.saving(ax, "systeme2.png")
```
