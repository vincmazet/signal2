# Filtrage : réponse fréquentielle de $e^{-t} u(t)$

```python
import matplotlib.pyplot as plt
import numpy as np

f = np.arange(0, 10, .01)
H = 1 / (1+1j*2*np.pi*f)
```

```python
fig, axs = plt.subplots(2,1, figsize=(10,10))
axs[0].plot(f, abs(H))
axs[0].grid()
axs[0].set_xlabel("$f (Hz)$")
axs[0].set_ylabel("$|H(f)|$")
axs[1].plot(f, np.angle(H))
axs[1].grid()
axs[1].set_xlabel("$f (Hz)$")
axs[1].set_ylabel("$Arg(H(f))$ (rad)")
plt.show()
```

```python
fig, axs = plt.subplots(2,1, figsize=(10,10))
axs[0].loglog(f, (abs(H)))
axs[0].grid()
axs[0].set_xlabel("$f (Hz)$")
axs[0].set_ylabel("$|H(f)|$ (dB)")
axs[1].semilogx(f, np.angle(H))
axs[1].grid()
axs[1].set_xlabel("$f (Hz)$")
axs[1].set_ylabel("$Arg(H(f))$ (rad)")
plt.show()
```
