# Figures pour la correction du TD

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('../../src/signal.mplstyle')
```

```python
# Échelon ou fonction de Heaviside

t = [-5, 0, 0, 5]
x = [0, 0, 1, 1]

fig, ax = plt.subplots(1,1, figsize=(10,3))
ax.plot(t,x)
ax.set_xlabel("$t$")
ax.set_ylabel("$u(t)$")
ax.set_xticks(range(-5,6))
ax.set_yticks(range(-5,6))
ax.axis('equal')
plt.savefig(f"echelon.svg")
plt.show()
```

```python
# Impulsion de Dirac

x = .1#.2
y = .9#.95
n = [-5, 0, 0, -x, 0, x, 0, 0, 5]
d = [ 0, 0, 1,  y, 1, y, 1, 0,  0]

fig, ax = plt.subplots(1,1, figsize=(10,3))
ax.plot(n,d)
ax.set_xlabel("$t$")
ax.set_ylabel("$\delta(t)$")
ax.set_xticks(range(-5,6))
ax.set_yticks(range(-5,6))
ax.axis('equal')
plt.savefig(f"dirac.svg")
plt.show()
```

```python
# Impulsion discrète

n = np.arange(-10, 10)
d = np.where(n==0, 1, 0)

plt.figure(figsize=(10,3))
plt.stem(n, d, basefmt=" ")
plt.xlabel("$n$")
plt.ylabel("$\delta[n]$")
ax.axis('equal')
plt.savefig(f"kronecker.svg")
plt.show()
```

```python
# Peigne de Dirac

# Flèche
x = .1
y = .9
arrow_x = np.array([0, 0, -x, 0, x, 0, 0])
arrow_y = np.array([0, 1,  y, 1, y, 1, 0])

# Affichage
fig, ax = plt.subplots(1,1, figsize=(10,3))
ax.plot([-5, 5], [0, 0])
T = 2
for i in range(-4, 6, T):
    ax.plot(arrow_x+i, arrow_y, color="C0")
ax.set_xlabel("$t$")
# ax.set_ylabel("$\cha(t)$")
ax.set_title(f"Peigne de période $T={T}$")
ax.set_xticks(range(-5,6))
ax.set_yticks(range(-5,6))
ax.axis('equal')
plt.savefig(f"peigne.svg")
plt.show()
```

```python
# x(t) = 4 sin(3 pi t + pi)

d = 2
fe = 100
Te = 1/fe
N = d/Te
n = np.arange(N)
t = n * Te
x = 4*np.sin(3*np.pi*t+np.pi)

# Affichage
fig, ax = plt.subplots(1,1, figsize=(10,3))
ax.plot(t, x)
ax.set_xlabel("$t$")
ax.set_ylabel("$x(t)$")
ax.set_title("$x(t) = 4\,\sin(3\pi t+\pi)$")
plt.savefig(f"signal-x.svg")
plt.show()
```

```python
# y(t) = 2 delta(t-1) - \delta(t+1)

# Flèche
x = .1
y = .9
arrow_x = np.array([0, 0, -x, 0, x, 0, 0])
arrow_y = np.array([0, 1,  y, 1, y, 1, 0])

# Affichage
fig, ax = plt.subplots(1,1, figsize=(10,3))
ax.plot([-4, 4], [0, 0], "C0")
ax.plot(arrow_x+1, 2*arrow_y, color="C0")
ax.plot(arrow_x-1, -arrow_y, color="C0")
ax.set_xlabel("$t$")
ax.set_ylabel("$y(t)$")
ax.set_title("$y(t) = 2\,\delta(t-1) - \delta(t+1)$")
plt.savefig(f"signal-y.svg")
plt.show()
```

```python
# Représentation schématique des produits de convolution

x = np.arange(-4, 4.1, 0.1)

fig = plt.figure(figsize=(5,2))
plt.plot([-4, -2, 0, 2, 4], [0, 0, 2, 0, 0], 'C0', linewidth=1.5)
plt.axis([-4, 4, -.5, 3])
fig.savefig(f"conv-1-result.svg", bbox_inches='tight')
plt.show()

fig = plt.figure(figsize=(5,2))
plt.plot(x, 2*np.exp(-(x-2)**2/0.5)+2*np.exp(-(x+2)**2/0.5), 'C0', linewidth=1.5)
plt.axis([-4, 4, -.5, 3])
fig.savefig(f"conv-2-result.svg", bbox_inches='tight')
plt.show()

fig = plt.figure(figsize=(5,2))
plt.plot([-4, -2.5, -1.5, 1.5, 2.5, 4],  [0, 0, 1, 1, 0, 0], 'C0', linewidth=1.5)
plt.axis([-4, 4, -.5, 3])
fig.savefig(f"conv-3-result.svg", bbox_inches='tight')
plt.show()
```

```python
# Spectre d'un cosinus

# Flèche
x = .1
y = .9
arrow_x = np.array([0, 0, -x, 0, x, 0, 0])
arrow_y = np.array([0, 1,  y, 1, y, 1, 0])

# Affichage
fig, ax = plt.subplots(1,1, figsize=(10,3))
ax.plot([-4, 4], [0, 0], "C0")
ax.plot(arrow_x+1, arrow_y/2, color="C0")
ax.plot(arrow_x-1, arrow_y/2, color="C0")
ax.set_xlabel("$f$")
ax.set_ylabel("$\mathcal{F}[\cos(t)]$")
plt.savefig(f"spectre-cos.svg")
plt.show()
```

```python
# Spectre d'une sinusoïde de fréquence 320 Hz est échantillonnée à 600 Hz.

xmin, xmax = -1600, 1600

def diracs(ax, xx, clr):
    # Affiche des Diracs aux abscisses xx

    # Ligne de base
    ax.plot([xmin, xmax], [0, 0], color=clr)

    for x in xx:
        dx = 20
        dy = .9
        arrow_x = np.array([-dx, 0, 0, 0, dx])
        arrow_y = np.array([ dy, 1, 0, 1, dy])
        ax.plot(arrow_x+x, arrow_y/2, color=clr)
        ax.plot(arrow_x-x, arrow_y/2, color=clr)

    ax.set_xlim([xmin, xmax])
    ax.set_xticks(np.arange(xmin, xmax+1, 200))

# Affichage
fig, axs = plt.subplots(3,1, figsize=(10,6))
fig.tight_layout(pad=2)

# Sinusoïde analogique à 320 Hz
diracs(axs[0], [-320, 320], "C0")
axs[0].annotate("320 Hz", xy=(320, 0), xytext=(400, 0.2), ha="left", arrowprops=dict(arrowstyle="->"))
axs[0].set_title("Sinusoïde analogique à 320 Hz")

# Sinusoïde échantillonnée à 600 Hz
fe = 600
for k in range(-3,4):
    xx = k * fe + np.array([-320, 320])
    diracs(axs[1], xx, "C0")
axs[1].annotate("320 Hz", xy=(320, 0), xytext=(400, 0.2), ha="left", arrowprops=dict(arrowstyle="->"))
axs[1].annotate("280 Hz", xy=(280, 0), xytext=(200, 0.2), ha="right", arrowprops=dict(arrowstyle="->"))
axs[1].set_title("Sinusoïde échantillonnée à 600 Hz")

# Signal reconstruit
diracs(axs[2], [-280, 280], "C0")
axs[2].annotate("280 Hz", xy=(280, 0), xytext=(200, 0.2), ha="right", arrowprops=dict(arrowstyle="->"))
axs[2].set_title("Signal reconstruit")
axs[2].set_xlabel("$f$")

plt.savefig(f"spectre-sampling.svg")
plt.show()
```
