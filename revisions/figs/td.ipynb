# Figures

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('../../src/signal.mplstyle')
```

## Sinus cardinaux

```python
x = np.arange(-3, 3, 0.05)

data = {
    "a": np.sinc(x),
    "b": .22 + np.sinc(x),
    "c": np.absolute(np.sinc(x)),
    "d": 1.3 * (-.12 + np.sinc(x))
}

fig = plt.figure(figsize=(10,6))

for i, name in enumerate(data):

    ax = plt.subplot(2, 2, i+1)
    plt.plot(x, data[name], 'C0', linewidth=1.5)
    plt.axis([-3, 3, -.5, 1.3])
    plt.text(-2.8, 1.2, name.upper(), fontsize=14, horizontalalignment='left', verticalalignment='top')
    plt.grid(True)

plt.savefig("sinc.svg")
plt.savefig("sinc.png")
plt.show()
```

## Signaux de convolution

```python
x = np.arange(-4, 4.1, 0.1)

data = {
    "dirac": [ [-4, 0, 0, 0, 4], [0, 0, 2, 0, 0], [0], [2] ],
    "diracs": [ [-4, -2, -2, -2, 2, 2, 2, 4], [0, 0, 2, 0, 0, 2, 0, 0], [-2, 2], [2, 2] ],
    "narrow-door": [ [-4, -.5, -.5, .5, .5, 4], [0, 0, 1, 1, 0, 0] ],
    "large-door": [ [-4, -2, -2, 2, 2, 4],  [0, 0, 1, 1, 0, 0] ],
    "triangle": [ [-4, -2, 0, 2, 4], [0, 0, 2, 0, 0] ],
    "gaussian": [ x, 2*np.exp(-x*x/0.5) ]
}

convs = [
    ["dirac", "triangle"],
    ["diracs", "gaussian"],
    ["narrow-door", "large-door"],
]

for j, c in enumerate(convs):
    
    fig = plt.figure(figsize=(10,2))
    for i, name in enumerate(c):

        ax = plt.subplot(1, 2, i+1)
        
        # Signal
        plt.plot(data[name][0], data[name][1], 'C0', linewidth=1.5)
        plt.axis([-4, 4, -.5, 3])
        plt.grid(True)
        
        # Flèche dez diracs éventuels
        if len(data[name])>2:
            plt.plot(data[name][2], data[name][3], 'C0^')
        
        # Produit de convolution
        if i==0:
            plt.text(4.5, 1, "*", fontsize=20)
        
    fig.savefig(f"conv-{j+1}.svg", bbox_inches='tight')
    fig.savefig(f"conv-{j+1}.png", bbox_inches='tight')
```
